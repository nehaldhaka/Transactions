```python
import pandas as pd  

df=pd.read_csv("transactions.csv")
df
```

```python
#1 What is the total sales amount across all transactions?

amount=df["t_amt"].sum() #using sum method to return total sales amount 
amount
```

```python
#2 Which month had the highest total transaction amount?


d = pd.to_datetime(df['t_date'],errors='coerce',dayfirst=True) #to convert it into datetime format (day first - it will return the days first)
t=d.dt.month_name() # coverting this t_date to month using month_name() method
df['month']=t #giving the name as month 
m=df.groupby('month')['t_amt'].sum().idxmax() #grouping them and returning the month that is having the highest total t_amt
m
df
#idxmax - returns the index of the maximum value 
```

```python
#3 What is the average transaction amount per customer?
df.groupby('cust_id')['t_amt'].mean()
```

```python
#4 What is the trend of total sales over months?
date=pd.to_datetime(df['t_date'],errors='coerce') #converting into date time format
df['t_amt']=pd.to_numeric(df['t_amt'],errors='coerce') #converting total amount to numeric

month=d.dt.to_period("M") #dt - it is used to extract or manipulate datetime properties from pandas
df['year_month']=month #converting monelty period for grouping 


sales=df.groupby('year_month')['t_amt'].sum().reset_index() #finding out the sales by grouping year and amount 
sales.sort_values('year_month')
```

```python
#5 What is the highest single transaction amount recorded?

df['t_amt']=pd.to_numeric(df['t_amt'],errors='coerce')
high=df['t_amt'].max()
high

```

```python
#6 Which service category contributes the most to total revenue?

rev=df.groupby('services')['t_amt'].sum()
rev.loc[[rev.idxmax()]] #loc will return the label based indexing 
```

```python
#7 Which product generated the highest revenue?

pro=df.groupby('products_used')['t_amt'].sum()
pro.loc[[pro.idxmax()]]
```

```python
#8 How does the average transaction amount vary between services?

df['t_amt']=pd.to_numeric(df['t_amt'],errors='coerce')
ser=df.groupby('services')['t_amt'].mean()
ser
```

```python
#9 How many unique customers made purchases?

df['cust_id'].nunique()
```

```python
#10 Which customers have spent the most overall?
cus=df.groupby('cust_id')['t_amt'].sum()
cus.loc[[cus.idxmax()]]
```

```python
#11 What is the average number of transactions per customer?

df.groupby('cust_id').size().mean() #size - counts the number of rows 
```

```python
#12 Are there customers who made multiple purchases in different categories?

mul=df.groupby('cust_id')['products_used'].nunique()
mul[mul>1]
```

```python
#13 What percentage of customers are repeat buyers?

((df.groupby('cust_id').size() > 1).mean()) * 100
```

```python
#14 Which product category has the highest total sales?

sale=df.groupby('products_used')['t_amt'].sum()
sale.loc[[sale.idxmax()]]
```

```python
#15 Which services are most popular by transaction count?

s=df.groupby('services').size()
s.loc[[s.idxmax()]]
```

```python
#16 For each service, which product type is purchased most frequently?

freq = df.groupby(['services', 'products_used']).size().reset_index(name='count')

freq.loc[freq.groupby('services')['count'].idxmax()]
```

```python
#19 Which state has the highest total sales?

state = df.groupby('state')['t_amt'].sum()
state.loc[[state.idxmax()]]
```

```python
#20 Which city records the highest number of transactions?

df.groupby('city').size().idxmax()
```

```python
#21 What is the average spending per transaction in each state?

df.groupby('state')['t_amt'].mean()
```

```python
#23 Which states buy the most Outdoor Recreation products?

out=df[df['products_used']=="Outdoor Recreation"]
count=out.groupby('state').size()
count.loc[[count.idxmax()]]
```
